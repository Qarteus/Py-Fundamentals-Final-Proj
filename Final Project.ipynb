```python
# Load dataset
flights = sns.load_dataset("flights")
```

```python
# Define label for total option
PassPerYear = "All Years"
```

```python
# Prepare dropdown options
def passenger_sorted_values(array):
    unique = array.unique().tolist()
    unique.sort()
    unique.insert(0, PassPerYear)
    return unique
```

```python
# Widget: dropdown for year selection
year_dropdown = widgets.Dropdown(
    options = passenger_sorted_values(flights['year']),
    description = 'Year:'
)
```

```python
# Output widget
output = widgets.Output()
```

```python
# Handler function
def handle_year_selection(year):
    with output:
        output.clear_output()
        if year == PassPerYear:
            total = flights['passengers'].sum()
            display(flights)
            print("Total passengers across all years: {}".format(total))
        else:
            filtered = flights[flights['year'] == year]
            total = filtered['passengers'].sum()
            display(filtered)
            print("Total passengers in {} : {}".format(year, total))
```

```python
# Connect widget and handler
interact(handle_year_selection, year = year_dropdown)

# Show output area
display(output)
```
